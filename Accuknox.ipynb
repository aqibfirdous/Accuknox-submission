#  Q1:
API Data Retrieval and Storage: You are tasked with fetching data from an external REST API, storing it in a local SQLite database, and displaying the retrieved data. The API provides a list of books in JSON format with attributes like title, author, and publication year.




```python
import requests
import sqlite3

# Fetch data from FREE Open Library API
url = "https://openlibrary.org/search.json?q=python"
response = requests.get(url)
data = response.json()

books = data["docs"][:10]
conn = sqlite3.connect("books.db")
cursor = conn.cursor()

print("Book data successfully stored in SQLite database.")
print("Database name: books.db")
print("Table name: books")

cursor.execute("SELECT title, author, year FROM books LIMIT 5")
rows = cursor.fetchall()

# Display the data
print("Books stored in database:\n")
for row in rows:
    title, author, year = row
    print(f"Title: {title}")
    print(f"Author: {author}")
    print(f"Publication Year: {year}")
    print("-" * 40)

conn.commit()
conn.close()

```

# Q2:
Data Processing and Visualization: Given a dataset containing information about students' test scores, fetch the data from an API, calculate the average score, and create a bar chart to visualize the data.


```python
import requests
import matplotlib.pyplot as plt
import random

# Fetch data from a free public API
url = "https://jsonplaceholder.typicode.com/posts"
response = requests.get(url).json()

names = [entry['title'] for entry in response[:10]]
scores = [random.randint(50, 100) for _ in names]
avg_score = sum(scores) / len(scores)
print("Average Score:", round(avg_score, 2))

# Plot bar chart
plt.figure(figsize=(12,8))
plt.bar(names, scores, color='skyblue')
plt.xticks(rotation=60, ha='right')
plt.ylabel("Scores")
plt.xlabel("Students (Post Titles)")
plt.title("Student Test Scores (Simulated based on Post Titles)")
plt.ylim(0, 100)
plt.tight_layout()
plt.show()
```

# Q3:
CSV Data Import to a Database: Write a Python script that reads data from a CSV file containing user information (e.g., name, email) and inserts it into a SQLite database.


```python
import csv
import sqlite3

conn = sqlite3.connect("users.db")
cursor = conn.cursor()

with open("users.csv", "r") as file:
    reader = csv.DictReader(file)
    for row in reader:
        cursor.execute(
            "INSERT INTO users (name, email) VALUES (?, ?)",
            (row["name"], row["email"])
        )
cursor.execute("SELECT * FROM users")
rows = cursor.fetchall()

for row in rows:
    print(row)
conn.commit()
conn.close()
```
